```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms

import json
import cv2
import numpy as np
```

# Create torch Dataset 

```python
class KeypointsDataset(Dataset):
    def __init__(self, img_dir, data_file):
        self.img_dir = img_dir
        with open(data_file, "r") as f:
            self.data = json.load(f)
        
        self.transforms = transforms.Compose([
            transforms.ToPILImage(),
            transforms.Resize((320, 320)),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        item = self.data[idx]
        img = cv2.imread(f"{self.img_dir}/{item['id']}.png")
        h,w = img.shape[:2]

        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = self.transforms(img)
        kps = np.array(item['kps']).flatten()
        kps = kps.astype(np.float32)

        kps[::2] *= 320.0 / w # Adjust x coordinates
        kps[1::2] *= 320.0 / h # Adjust y coordinates
        kps[::2] = np.clip(kps[::2], 0, 319)
        kps[1::2] = np.clip(kps[1::2], 0, 319)

        return img, kps
```

```python
train_dataset = KeypointsDataset("data/images","data/data_train.json")
val_dataset = KeypointsDataset("data/images","data/data_val.json")

train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=8, shuffle=True)
```

Creat model

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = models.resnet50(pretrained=True)
model.fc =  torch.nn.Linear(model.fc.in_features, 14*2) # Replaces the last layer
model = model.to(device)
criterion = torch.nn.MSELoss()
```

```python
stage1_train_losses = []
stage1_val_losses = []

stage2_train_losses = []
stage2_val_losses = []
```

Train model

```python
for param in model.parameters():
    param.requires_grad = False

for param in model.fc.parameters():
    param.requires_grad = True

optimizer = torch.optim.Adam(model.fc.parameters(), lr=1e-3)
```

```python
epochs_stage1 = 8

for epoch in range(epochs_stage1):
    model.train()
    running_loss = 0

    for i, (imgs, kps) in enumerate(train_loader):
        imgs = imgs.to(device)
        kps = kps.to(device)

        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, kps)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        if i % 10 == 0:
            print(f"[Stage1] Epoch {epoch+1}/{epochs_stage1}, Iter {i}/{len(train_loader)}, Loss: {loss.item():.4f}")

    avg_train_loss = running_loss / len(train_loader)
    stage1_train_losses.append(avg_train_loss)

    # Validation
    model.eval()
    val_running_loss = 0

    with torch.no_grad():
        for imgs, kps in val_loader:
            imgs = imgs.to(device)
            kps = kps.to(device)

            outputs = model(imgs)
            loss = criterion(outputs, kps)
            val_running_loss += loss.item()

    avg_val_loss = val_running_loss / len(val_loader)
    stage1_val_losses.append(avg_val_loss)

    print(f"[Stage1] Epoch {epoch+1} Average Train Loss: {avg_train_loss:.4f}, Average Val Loss: {avg_val_loss:.4f}")
```

```python
for param in model.layer4.parameters():
    param.requires_grad = True

optimizer = torch.optim.Adam([
    {"params": model.layer4.parameters(), "lr": 1e-4},
    {"params": model.fc.parameters(), "lr": 5e-4},
])
```

```python
epochs_stage2 = 15

for epoch in range(epochs_stage2):
    model.train()
    running_loss = 0

    for i, (imgs, kps) in enumerate(train_loader):
        imgs = imgs.to(device)
        kps = kps.to(device)

        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, kps)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        if i % 10 == 0:
            print(f"[Stage2] Epoch {epoch+1}/{epochs_stage2}, Iter {i}/{len(train_loader)}, Loss: {loss.item():.4f}")

    avg_train_loss = running_loss / len(train_loader)
    stage2_train_losses.append(avg_train_loss)

    # Validation
    model.eval()
    val_running_loss = 0

    with torch.no_grad():
        for imgs, kps in val_loader:
            imgs = imgs.to(device)
            kps = kps.to(device)

            outputs = model(imgs)
            loss = criterion(outputs, kps)
            val_running_loss += loss.item()

    avg_val_loss = val_running_loss / len(val_loader)
    stage2_val_losses.append(avg_val_loss)

    print(f"[Stage2] Epoch {epoch+1} Average Train Loss: {avg_train_loss:.4f}, Average Val Loss: {avg_val_loss:.4f}")
```

```python
import matplotlib.pyplot as plt

# -------------------
# Plot Stage 1
# -------------------
epochs1 = list(range(1, len(stage1_train_losses) + 1))

plt.figure(figsize=(8, 5))
plt.plot(epochs1, stage1_train_losses, marker='o', label='Training Loss')
plt.plot(epochs1, stage1_val_losses, marker='s', label='Validation Loss')

plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Stage 1 Training and Validation Loss Curve (Keypoint Detection)')
plt.legend()
plt.grid(True)
plt.savefig('stage1_keypoint_train_val_loss.png', dpi=300, bbox_inches='tight')
plt.show()

# -------------------
# Plot Stage 2
# -------------------
epochs2 = list(range(1, len(stage2_train_losses) + 1))

plt.figure(figsize=(8, 5))
plt.plot(epochs2, stage2_train_losses, marker='o', label='Training Loss')
plt.plot(epochs2, stage2_val_losses, marker='s', label='Validation Loss')

plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Stage 2 Training and Validation Loss Curve (Keypoint Detection)')
plt.legend()
plt.grid(True)
plt.savefig('stage2_keypoint_train_val_loss.png', dpi=300, bbox_inches='tight')
plt.show()
```
